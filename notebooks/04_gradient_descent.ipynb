## 📚 Syllabus Concepts Covered:
1. **Batch Gradient Descent (BGD):**
   Computes the exact gradient of the loss function over the **entire training dataset** ($m$ samples) per step:
   $$\mathbf{w} := \mathbf{w} - \alpha \cdot \frac{1}{m} \sum_{i=1}^m (\hat{y}^{(i)} - y^{(i)}) \mathbf{x}^{(i)}$$
   - *Pros:* Stable, guaranteed convergence to global optimum for convex loss.
   - *Cons:* Computationally expensive on large flight datasets.

2. **Stochastic Gradient Descent (SGD):**
   Updates parameters after evaluating **a single sample** ($i$) at each step:
   $$\mathbf{w} := \mathbf{w} - \alpha \cdot (\hat{y}^{(i)} - y^{(i)}) \mathbf{x}^{(i)}$$
   - *Pros:* Extremely fast, highly scalable to millions of flights.
   - *Cons:* Noisy gradient fluctuations; requires decaying learning rate schedule.

3. **Mini-Batch Gradient Descent (MBGD):**
   Evaluates a small batch of $B$ samples (e.g. $B = 64$ or $128$):
   $$\mathbf{w} := \mathbf{w} - \alpha \cdot \frac{1}{B} \sum_{i \in \mathcal{B}} (\hat{y}^{(i)} - y^{(i)}) \mathbf{x}^{(i)}$$
   - *Pros:* Balances computational vectorization efficiency with stochastic exploration.

4. **Convergence Analysis:**
   Plotting cost function $J(\mathbf{w})$ vs. iteration number across different learning rates $\alpha \in \{0.001, 0.01, 0.1\}$.
5. **Comparison with Scikit-Learn's `SGDClassifier`:**
   Benchmarking scratch implementations against production SGD.

```python
# Step 2: First-Principles Gradient Descent Architectures
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -250, 250)))

def compute_loss(X, y, w, b):
    m = len(y)
    p = sigmoid(np.dot(X, w) + b)
    eps = 1e-15
    p = np.clip(p, eps, 1 - eps)
    return - (1.0 / m) * np.sum(y * np.log(p) + (1 - y) * np.log(1 - p))

print("✓ Core vectorized gradient computation functions defined.")
print("  - Sigmoid activation with numerical clipping")
print("  - Binary cross-entropy loss function ready for BGD, SGD, and MBGD benchmarking.")
```
