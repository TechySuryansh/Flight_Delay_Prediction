## 📚 Syllabus Concepts Covered:
1. **Confusion Matrix Breakdown:**
   - **True Positives (TP):** Correctly predicted delayed flights.
   - **False Positives (FP) [Type I Error]:** On-time flights falsely predicted as delayed (unnecessary alerts).
   - **False Negatives (FN) [Type II Error]:** Delayed flights missed by the model (costly disruptions).
   - **True Negatives (TN):** Correctly predicted on-time flights.

2. **Why Accuracy is Misleading (The Class Imbalance Problem):**
   In flight data with ~80% on-time and ~20% delayed flights:
   $$\text{Dummy Accuracy} = \frac{0.80 \times N + 0}{N} = 80\% \quad (\text{Recall} = 0\%!)$$
   Accuracy fails to penalize complete blindness to delays.

3. **Classification Metrics:**
   - **Precision:** $\frac{\text{TP}}{\text{TP} + \text{FP}}$ (How trustworthy is a delay alert?)
   - **Recall (Sensitivity):** $\frac{\text{TP}}{\text{TP} + \text{FN}}$ (What fraction of delayed flights did we catch?)
   - **Specificity:** $\frac{\text{TN}}{\text{TN} + \text{FP}}$ (Correct identification of on-time flights)
   - **F1-Score:** $2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$ (Harmonic mean)

4. **Curves and Threshold Moving:**
   - **ROC Curve & ROC-AUC:** Trade-off between True Positive Rate and False Positive Rate across all thresholds.
   - **PR Curve & PR-AUC:** Crucial metric under class imbalance, evaluating precision across varying recall levels.
   - **Operational Threshold Selection:** Tuning threshold $\tau$ away from standard $0.5$ (e.g. to $0.35$ or $0.40$) to prioritize early disruption prevention.
