## 📚 Syllabus Concepts Covered:
1. **Binary Classification & The Sigmoid Function:**
   $$P(y=1|\mathbf{x}) = \sigma(z) = \frac{1}{1 + e^{-z}}, \quad \text{where } z = \mathbf{w}^T \mathbf{x} + b$$
2. **Maximum Likelihood Estimation (MLE):**
   Likelihood function $L(\mathbf{w}) = \prod_{i=1}^m p_i^{y^{(i)}} (1 - p_i)^{1 - y^{(i)}}$.
   Minimizing the negative log-likelihood yields the **Log-Loss / Binary Cross-Entropy Cost Function**:
   $$J(\mathbf{w}, b) = -\frac{1}{m} \sum_{i=1}^m \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]$$
3. **Regularization Techniques:**
   - **L1 Regularization (Lasso):** Adds penalty $\lambda \sum |w_j|$, driving irrelevant weights to exact zero (Feature Selection).
   - **L2 Regularization (Ridge):** Adds penalty $\frac{\lambda}{2} \sum w_j^2$, shrinking weights toward zero and preventing extreme coefficients.
4. **Bias-Variance Tradeoff:**
   Tuning regularization strength parameter $C = \frac{1}{\lambda}$. Large $C$ (low regularization) risks high variance / overfitting; small $C$ (high regularization) risks high bias / underfitting.
5. **Model Interpretation:**
   Examining learned regression weights $\mathbf{w}$ and odds ratios $e^{w_j}$ to determine which pre-departure features increase or decrease delay risk.
